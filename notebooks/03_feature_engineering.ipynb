# Feature Engineering
Compute simple statistics and an FFT magnitude visualization.

```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.append(str(ROOT))

from src.data.loader import load_mat_signal
from src.processing.features import simple_stats

sample = ROOT / "data" / "36" / "1311" / "data.mat"
signal = load_mat_signal(sample, key="dataMeasured1")
raw = signal.real if np.iscomplexobj(signal) else signal

stats = simple_stats(raw)
print("mean, std, max, min:", stats)

fft_mag = np.abs(np.fft.rfft(raw))
fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(fft_mag)
ax.set_title("FFT Magnitude")
ax.set_xlabel("Frequency Bin")
ax.set_ylabel("Magnitude")
fig.tight_layout()

out = ROOT / "outputs" / "figures" / "spectrograms" / "fft_magnitude_nb.png"
out.parent.mkdir(parents=True, exist_ok=True)
fig.savefig(out)
plt.close(fig)
print("Saved FFT magnitude plot to outputs/figures/spectrograms.")
```
